## 2. Lire et nettoyer les fichiers

````python
import pandas as pd, io, subprocess
from pathlib import Path

airlines = pd.read_json('airlines.json')
airports = pd.read_excel('airports.xlsx').drop(columns=['Unnamed: 0'], errors='ignore')
raw_flights = pd.read_excel('flights.xlsx', dtype=str)
assert raw_flights.shape[1] == 1, 'Format flights inattendu'
flight_csv = raw_flights.columns[0] + '\n' + '\n'.join(raw_flights.iloc[:,0].fillna('').astype(str))
flights = pd.read_csv(io.StringIO(flight_csv), low_memory=False)
planes = pd.read_html('planes.html')[0].drop(columns=['Unnamed: 0'], errors='ignore')
result = subprocess.run(['pdftotext', '-layout', 'weather.pdf', '-'], capture_output=True, text=True, check=True)
lines = [s.strip() for s in result.stdout.splitlines() if s.strip()]
header = 'origin,year,month,day,hour,temp,dewp,humid,wind_dir,wind_speed,wind_gust,precip,pressure,visib,time_hour'
assert header in lines, 'En-tête météo introuvable dans le PDF'
weather_lines = [header] + [s for s in lines if s != header and s.count(',') == 14 and not s.startswith('origin,')]
weather = pd.read_csv(io.StringIO('\n'.join(weather_lines)), skipinitialspace=True)

frames = {'airlines': airlines, 'airports': airports, 'flights': flights, 'planes': planes, 'weather': weather}
expected = {
'airlines': ['carrier','name'],
'airports': ['faa','name','lat','lon','alt','tz','dst','tzone'],
'flights': ['year','month','day','dep_time','sched_dep_time','dep_delay','arr_time','sched_arr_time','arr_delay','carrier','flight','tailnum','origin','dest','air_time','distance','hour','minute','time_hour'],
'planes': ['tailnum','year','type','manufacturer','model','engines','seats','speed','engine'],
'weather': header.split(',')
}
for name, df in frames.items():
    assert list(df.columns) == expected[name], f'Colonnes inattendues pour {name}: {df.columns.tolist()}'
    print(f'{name}: {len(df):,} lignes; {len(df.columns)} colonnes')
````

Le script vide d’abord les cinq tables pour rendre la réimportation reproductible. Ne l’exécuter que sur les tables du projet destinées à être remplacées.

````python
fichier = io.StringIO()
flights.to_csv(fichier, index=False, header=False, na_rep="")
fichier.seek(0)

with psycopg.connect(url, sslmode="require") as conn:
    with conn.cursor() as cur:
        with cur.copy("""
            COPY public.flights
            (year, month, day, dep_time, sched_dep_time,
             dep_delay, arr_time, sched_arr_time, arr_delay,
             carrier, flight, tailnum, origin, dest,
             air_time, distance, hour, minute, time_hour)
            FROM STDIN WITH (FORMAT CSV, NULL '')
        """) as copy:
            copy.write(fichier.getvalue())

print("Importation des vols terminée !")
````

````python
import psycopg, csv
from google.colab import userdata

url = userdata.get('DATABASE_URL')
assert url and url.startswith(('postgresql://', 'postgres://')), 'Secret DATABASE_URL manquant ou invalide'
# Les connexions Supabase PostgreSQL exigent TLS.
from urllib.parse import urlsplit, parse_qsl, urlencode, urlunsplit
parts = urlsplit(url)
qs = dict(parse_qsl(parts.query))
qs['sslmode'] = 'require'
url = urlunsplit((parts.scheme, parts.netloc, parts.path, urlencode(qs), parts.fragment))

with psycopg.connect(url, connect_timeout=20) as conn:
    with conn.cursor() as cur:
        # Transaction : en cas d'erreur, tout est annulé.
        cur.execute('TRUNCATE TABLE public.flights, public.weather, public.planes, public.airports, public.airlines RESTART IDENTITY')
        for name, df in frames.items():
            cols = expected[name]
            colsql = ', '.join('"'+c+'"' for c in cols)
            statement = f'COPY public."{name}" ({colsql}) FROM STDIN WITH (FORMAT CSV, NULL "")'
            with cur.copy(statement) as cp:
                for start in range(0, len(df), 10000):
                    chunk = df.iloc[start:start+10000]
                    buf = io.StringIO()
                    chunk.to_csv(buf, index=False, header=False, na_rep='', quoting=csv.QUOTE_MINIMAL)
                    cp.write(buf.getvalue())
            cur.execute(f'SELECT COUNT(*) FROM public."{name}"')
            print(f'{name}: {cur.fetchone()[0]:,} lignes importées')
    conn.commit()
print('Import terminé.')
````
